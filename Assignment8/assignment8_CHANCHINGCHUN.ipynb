## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

age = df["age"].dropna()
age_male = df[df["sex"] == "male"]["age"].dropna()
age_female = df[df["sex"] == "female"]["age"].dropna()

fig, axs = plt.subplots(1, 2, figsize=(14,5))

bin_widths = ['3','3']
labels=['男性乗客', '女性乗客']

bins= np.arange(0, 82, 1.6)

#男性
axs[0].hist(age, bins=bins, color = "gray", label = "すべての乗客")
axs[0].hist(age_male, bins=bins, color = "blue", label = "男性乗客")
axs[0].set_title('男性乗客')
axs[0].set_xlabel("年齢(歳)")
axs[0].set_ylabel("人数")
axs[0].grid(axis='y', alpha=0.3)
axs[0].set_xticks([0, 10, 20, 30, 40, 50, 60, 70, 80])
axs[0].set_yticks([0, 10, 20, 30, 40, 50, 60, 70, 80])
axs[0].minorticks_on()
axs[0].tick_params(which='major', direction='in', length=6, width=1.5,top=True, right=True)
axs[0].tick_params(which='minor', direction='in', length=3, width=1,top=True, right=True)
axs[0].legend()

#女性
axs[1].hist(age, bins=bins, color="gray", label = "すべての乗客")
axs[1].hist(age_female, bins=bins, color="red", label = "女性乗客")
axs[1].set_title('女性乗客')
axs[1].set_xlabel("年齢(歳)")
axs[1].set_ylabel('人数')
axs[1].grid(axis='y', alpha=0.3)
axs[1].set_xticks([0, 10, 20, 30, 40, 50, 60, 70, 80])
axs[1].set_yticks([0, 10, 20, 30, 40, 50, 60, 70, 80])
axs[1].minorticks_on()
axs[1].tick_params(which='major', direction='in', length=6, width=1.5,top=True, right=True)
axs[1].tick_params(which='minor', direction='in', length=3, width=1,top=True, right=True)
axs[1].legend()
```
